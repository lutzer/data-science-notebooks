## 2. Import core libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print(f"numpy      {np.__version__}")
print(f"pandas     {pd.__version__}")
print(f"matplotlib {plt.matplotlib.__version__}")
```

## 3. Build a small DataFrame

```python
rng = np.random.default_rng(seed=42)
df = pd.DataFrame({
    "x": np.linspace(0, 2 * np.pi, 100),
})
df["sin"] = np.sin(df["x"])
df["noisy"] = df["sin"] + rng.normal(0, 0.1, size=len(df))
df.head()
```

## 4. Plot it

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(df["x"], df["sin"], label="sin(x)")
ax.scatter(df["x"], df["noisy"], s=10, alpha=0.5, label="noisy samples")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
plt.show()
```
