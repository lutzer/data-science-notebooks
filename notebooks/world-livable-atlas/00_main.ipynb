# 00 - Main

Run this script to execute all other scripts sequenctially. Check `output` folder if errors occur.


```python
import papermill as pm
from pathlib import Path

notebooks = [
    "01_setup.ipynb",
    "02_grid.ipynb",
    "03_region_mask.ipynb",
    "11_sea_proximity.ipynb",
    "12_terrain_ruggedness.ipynb",
    "13_sun_hours.ipynb",
    "14_temperature_pleasantness.ipynb",
    "15_annual_greenness.ipynb",
    "16_precipitation_balance.ipynb",
    "17_air_quality.ipynb",
    "18_natural_disaster_risk.ipynb",
    "19_climate_vulnerability.ipynb",
    "21_population_density.ipynb",
    "22_urbanity.ipynb",
    "23_internet_connectivity.ipynb",
    "24_healthcare_access.ipynb",
    "31_income.ipynb",
    "32_cost_of_living.ipynb",
    "33_crime_rate.ipynb",
    "34_human_freedom.ipynb",
    "35_corruption.ipynb",
    "36_education.ipynb",
    "37_people_happiness.ipynb"
]
output_dir = Path("output")
output_dir.mkdir(parents=True, exist_ok=True)

for nb in notebooks:
    print(f"Running notebook: {nb}")
    pm.execute_notebook(
        input_path=nb,
        output_path=str(output_dir / nb.replace(".ipynb", "_output.ipynb")),
        extra_arguments=["--log-level=ERROR"],
    )

print("All notebooks ran successfully.")
```
